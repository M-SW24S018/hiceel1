#Бодлого
Гараас өгсөн list доторх элементийн тоолдог код бич

```python
arr = list(map(int, input().split()))

count_dict = {}
for el in arr:
  if el in count_dict:
    count_dict[el] +=1
else:
  count_dict[el] = 1

print(count_dict)

```

```python
for  k in count_dict:
  print('{} элемент {} удаа орсон байна.'. format(k, count_dict[k]))
```
